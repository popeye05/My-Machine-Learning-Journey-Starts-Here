# Experiment 3C
## Submitted by Swarnab Das. BTECH/CSE/24/080

```python
print("Submitted by Swarnab Das/nBTECH/CSE/24/080")
plt.figure(figsize=(8,5)) # This is the first thing i.e setting the canvas
plt.plot(
    df["Roll_No"],
    df["Marks"],
    marker = 'o'
)
plt.title("Students Marks Analysis", y= -0.25)
#Labelling the axes
plt.xlabel("Student Roll NUmber")
plt.ylabel("Student Marks")
highest = df["Marks"].idxmax()
#Here We add the annotation
plt.annotate(
    "Highest Marks",
    xy=(
        df.loc[highest,"Roll_No"],
        df.loc[highest,"Marks"]
    ),
    xytext=(
        df.loc[highest,"Roll_No"]+2,
        df.loc[highest,"Marks"] +5
    ),
    arrowprops =dict(facecolor="red")
)
# Creating Grid Lines
plt.grid(True)
plt.show()
```

# Experiment 3D
## Submitted by Swarnab Das. BTECH/CSE/24/080

```python
print("Submitted by Swarnab Das/nBTECH/CSE/24/080")
plt.figure(figsize=(8,5)) # This is the first thing i.e setting the canvas
plt.plot(
    df["Roll_No"],
    df["Marks"],
    marker = 'o'
)
plt.title("Students Marks Analysis", y= -0.25)
#Labelling the axes
plt.xlabel("Student Roll NUmber")
plt.ylabel("Student Marks")
highest = df["Marks"].idxmax()
lowest = df["Marks"].idxmin()
#Here We add the annotation
plt.annotate(
    "Highest Marks",
    xy=(
        df.loc[highest,"Roll_No"],
        df.loc[highest,"Marks"]
    ),
    xytext=(
        df.loc[highest,"Roll_No"]+2,
        df.loc[highest,"Marks"] +5
    ),
    arrowprops =dict(facecolor="red")
)
plt.annotate(
    "Lowest Marks",
    xy=(
        df.loc[lowest,"Roll_No"],
        df.loc[lowest,"Marks"]
    ),
    xytext=(
        df.loc[lowest,"Roll_No"]+1,
        df.loc[lowest,"Marks"] -10
    ),
    arrowprops =dict(facecolor="blue")
)
# Creating Grid Lines
plt.grid(True)
plt.show()
```

# Experiment 3I

```python
#Finding The Correlation
corr = df.corr(numeric_only = True)
plt.figure(figsize=(8 ,6))
# Display color map
plt.imshow(
    corr,
    cmap = "viridis",
    )

plt.colorbar()
plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)
plt.yticks(
    range(len(corr.columns)),
    corr.columns,
)
plt.title("Correlation Matrix")
plt.show()
```
