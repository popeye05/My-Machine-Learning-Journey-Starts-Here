```python
import pandas as pd
import numpy as np
#Experiment Done By SWARNAB DAS BTECH/CSE/24/080
# Set random seed for reproducibility
np.random.seed(42)

# Number of students
n = 20

# Create synthetic dataset
student_data = pd.DataFrame({
    "Roll_No": np.arange(101, 101 + n),
    "Name": [f"Student_{i}" for i in range(1, n + 1)],
    "Age": np.random.randint(18, 24, n),
    "Gender": np.random.choice(["Male", "Female"], n),
    "Department": np.random.choice(["CSE ", "IT", "ECE ", "EE", "ME"], n),
    "Semester": np.random.randint(1, 9, n),
    "Attendance (%)": np.random.randint(60, 101, n),
    "Marks": np.random.randint(35, 101, n),
    "Study_Hours": np.round(np.random.uniform(1, 8, n), 1),
    "Sports": np.random.choice(["Yes ", "No"], n)
})

# ------------------------------------------------
# adding Missing Values
# ------------------------------------------------

student_data.loc[2, "Age "] = np.nan
student_data.loc[5, " Marks "] = np.nan
student_data.loc[7, " Department "] = np.nan
student_data.loc[10, " Attendance (%)"] = np.nan
student_data.loc[14, " Study_Hours "] = np.nan
student_data.loc[18, " Gender "] = np.nan

# ------------------------------------------------
# adding Duplicate Records
# ------------------------------------------------

student_data = pd.concat(
    [student_data, student_data.iloc[[3]], student_data.iloc[[8]]],
    ignore_index=True
)

# Save the dataset
student_data.to_csv("student_data_dirty.csv", index=False)

# Display dataset
print("Dataset with Missing Values and Duplicate Records")
print(student_data)

print("\nDataset Shape :", student_data.shape)

print("\nDataset saved as ’student_data_dirty.csv ’")
```

```python
print("EXPERIMENT BY SWARNAB DAS BTECH/CSE/24/080")
import pandas as pd

# Read the dataset
data = pd.read_csv("student_data_dirty.csv")

# Display the first five records
print("========== ORIGINAL DATASET ========== ")
print(data.head())

# Check for missing values
print("\n========== MISSING VALUES ========== ")
print(data.isnull())

# Count missing values in each column
print("\n========== TOTAL MISSING VALUES ========== ")
print(data.isnull().sum())

# Check for duplicate rows
print("\n========== DUPLICATE RECORDS ========== ")
print(data.duplicated().sum())

# Remove duplicate rows
data = data.drop_duplicates()

# Fill missing numerical values with the column mean
data["Marks"] = data["Marks"].fillna(data["Marks"].mean())
data["Attendance (%)"] = data["Attendance (%)"].fillna(data["Attendance (%)"].mean())
data["Study_Hours"] = data["Study_Hours"].fillna(data["Study_Hours"].mean())

# Fill missing categorical values with the mode
data["Department"] = data["Department"].fillna(data["Department"].mode()[0])
data["Gender"] = data["Gender"].fillna(data["Gender"].mode()[0])
data["Sports"] = data["Sports"].fillna(data["Sports"].mode()[0])

# Rename column
data.rename(columns={"Attendance (%)": "Attendance"}, inplace=True)

# Convert data type
data["Semester"] = data["Semester"].astype(int)

# Display cleaned dataset
print("\n========== CLEANED DATASET ========== ")
print(data.head())

# Display dataset information
print("\n========== DATASET INFORMATION ========== ")
data.info()

# Save the cleaned dataset
data.to_csv("student_data.csv", index=False)

print("\nCleaned dataset has been saved as student_data.csv")
```

```python
# ============================================================
# EXPERIMENT 1d : BAR CHART USING MATPLOTLIB
# ============================================================
print("EXPERIMENT BY SWARNAB DAS BTECH/CSE/24/080")
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt

# Read the dataset
data = pd.read_csv("student_data.xls")

# Calculate average marks department - wise
average_marks = data.groupby("Department")["Marks"].mean()

# Display the calculated average marks
print("Average Marks of Each Department ")
print(average_marks)

# Create the bar chart
plt.figure(figsize=(8,5))
plt.bar(average_marks.index,
        average_marks.values,
        color=['red','blue','green','orange','purple'])

# Add chart title
plt.title("Average Marks of Students by Department")

# Add X- axis label
plt.xlabel("Department")

# Add Y- axis label
plt.ylabel("Average Marks")

# Display values on top of each bar
for i, value in enumerate(average_marks.values):
    plt.text(i, value + 0.5, f"{value:.1f}", ha='center')

# Display grid
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Display the chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
print("EXPERIMENT BY SWARNAB DAS BTECH/CSE/24/080")
# Read the dataset
data = pd.read_csv("student_data.xls")

# Count the number of students in each department
department_count = data["Department"].value_counts()

# Display the department-wise student count
print("Number of Students in Each Department")
print(department_count)

# Create the pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    department_count.values,
    labels=department_count.index,
    autopct='%1.1f%%',
    startangle=90,
    shadow=True
)

# Add chart title
plt.title("Distribution of Students by Department")

# Display the chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

print("EXPERIMENT BY SWARNAB DAS BTECH/CSE/24/080")
# Read the dataset
data = pd.read_csv("student_data.csv")

# Display the first five records
print("First Five Records")
print(data.head())

# Display descriptive statistics of Marks
print("\nStatistics of Marks")
print(data["Marks"].describe())

# Create the box plot
plt.figure(figsize=(6, 6))

plt.boxplot(
    data["Marks"],
    patch_artist=True,
    labels=["Marks"],
    boxprops=dict(facecolor="lightgreen"),
    medianprops=dict(color="red", linewidth=2),
    whiskerprops=dict(color="blue"),
    capprops=dict(color="black"),
    flierprops=dict(marker='o',
                    markerfacecolor='red',
                    markersize=8)
)

# Add chart title
plt.title("Box Plot of Student Marks")

# Add Y-axis label
plt.ylabel("Marks")

# Display grid
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Display the chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

print("EXPERIMENT BY SWARNAB DAS BTECH/CSE/24/080")
# Read the dataset
data = pd.read_csv("student_data.xls")

# Display the first five records
print("First Five Records")
print(data.head())

# Display descriptive statistics
print("\nDescriptive Statistics")
print(data[["Marks", "Attendance (%)", "Study_Hours"]].describe())

# Create the figure
plt.figure(figsize=(9, 6))

# Create multiple box plots
plt.boxplot(
    [
        data["Marks"],
        data["Attendance (%)"],
        data["Study_Hours"]
    ],
    labels=["Marks", "Attendance (%)", "Study Hours"],
    patch_artist=True,
    boxprops=dict(facecolor="lightblue"),
    medianprops=dict(color="red", linewidth=2),
    whiskerprops=dict(color="blue"),
    capprops=dict(color="black"),
    flierprops=dict(marker='o',
                    markerfacecolor='red',
                    markersize=7)
)

# Add chart title
plt.title("Comparison of Marks, Attendance and Study Hours")

# Add Y-axis label
plt.ylabel("Values")

# Display grid
plt.grid(axis='y', linestyle='--', alpha=0.5)

# Display the chart
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.tri as tri

print("EXPERIMENT BY SWARNAB DAS BTECH/CSE/24/080")
# Read the dataset
data = pd.read_csv("student_data.xls")

# Display the first five records
print("First Five Records")
print(data.head())

# Select X and Y coordinates
x = data["Study_Hours"]
y = data["Marks"]

# Create triangulation
triangulation = tri.Triangulation(x, y)

# Create figure
plt.figure(figsize=(8, 6))

# Draw triangulation mesh
plt.triplot(triangulation, color='blue')

# Display original data points
plt.scatter(x, y, color='red')

# Add chart title
plt.title("Triangulation Plot of Study Hours vs Marks")

# Add X-axis label
plt.xlabel("Study Hours")

# Add Y-axis label
plt.ylabel("Marks")

# Display grid
plt.grid(True)

# Display the graph
plt.show()
```
